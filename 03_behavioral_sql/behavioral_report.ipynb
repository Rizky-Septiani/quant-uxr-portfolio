````python
# ============================================================
# CELL 1 — SQLite Environment Setup in Google Colab
# Objective: Initialize an in-memory SQL database engine directly
# within Python to execute relational behavioral data extraction.
# (Industry standard databases like PostgreSQL or BigQuery share ~95% identical syntax).
# ============================================================

import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import warnings
warnings.filterwarnings("ignore")

# Initialize an in-memory SQLite database connection
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

# Establish corporate design palette for behavioral visualizations
COLORS = {
    "primary" : "#2C3E50",
    "accent"  : "#3498DB",
    "success" : "#27AE60",
    "warning" : "#F39C12",
    "danger"  : "#E74C3C",
    "light"   : "#ECF0F1",
    "neutral" : "#95A5A6",
    "purple"  : "#8E44AD",
    "teal"    : "#16A085"
}

# Configure global plotting aesthetics
plt.rcParams.update({
    "figure.facecolor" : "white",
    "axes.facecolor"   : "white",
    "axes.spines.top"  : False,
    "axes.spines.right": False,
    "axes.titlesize"   : 12,
    "axes.titleweight" : "bold",
    "axes.labelsize"   : 10,
    "font.family"      : "sans-serif",
    "figure.dpi"       : 120
})

# Helper function: Execute SQL queries and format output as a clean pandas DataFrame
def sql(query, show_rows=10):
    """
    Executes a SQL query against the in-memory database and returns a formatted DataFrame.
    Streamlines data retrieval without re-writing pd.read_sql_query for every transaction.
    """
    result = pd.read_sql_query(query, conn)
    if show_rows:
        print(result.head(show_rows).to_string(index=False))
    return result

print("✅ SQLite Environment Successfully Initialized!")
print(f"   Python Version: {sys.version.split()[0]} | SQLite Version: {sqlite3.sqlite_version}")
````

````python
# ============================================================
# CELL 2 — Constructing the GajiKu Product Analytics Database Architecture
# Objective: Generate a 4-table relational database model simulating
# realistic user behavior, event telemetries, session logs, and financial transactions.
# ============================================================

np.random.seed(2024)

# ── TABLE 1: users ───────────────────────────────────────────
# Master user directory capturing demographic and acquisition profiles
n_users = 500
users_data = pd.DataFrame({
    "user_id"          : [f"USR{i:05d}" for i in range(1, n_users+1)],
    "name"             : [f"User_{i}" for i in range(1, n_users+1)],
    "user_type"        : np.random.choice(
                             ["new", "returning"], n_users,
                             p=[0.45, 0.55]),
    "city"             : np.random.choice(
                             ["Jakarta", "Bandung", "Surabaya", "Medan"],
                             n_users, p=[0.50, 0.20, 0.20, 0.10]),
    "device"           : np.random.choice(
                             ["Android", "iOS"], n_users,
                             p=[0.63, 0.37]),
    "channel"          : np.random.choice(
                             ["Organic", "Referral", "Paid Ad"],
                             n_users, p=[0.35, 0.40, 0.25]),
    "registration_date": pd.date_range(
                             "2023-01-01", "2024-03-31",
                             periods=n_users
                         ).strftime("%Y-%m-%d"),
    "plan"             : np.random.choice(
                             ["free", "premium"], n_users,
                             p=[0.70, 0.30])
})

# ── TABLE 2: sessions ────────────────────────────────────────
# Captures application launch logs and session-level engagement metrics
n_sessions = 8000
session_users = np.random.choice(
    users_data["user_id"], n_sessions,
    p=None  # Uniform distribution across user base
)
sessions_data = pd.DataFrame({
    "session_id"     : [f"SES{i:06d}" for i in range(1, n_sessions+1)],
    "user_id"        : session_users,
    "session_date"   : pd.date_range(
                           "2024-01-01", "2024-03-31",
                           periods=n_sessions
                       ).strftime("%Y-%m-%d"),
    "duration_sec"   : np.abs(
                           np.random.normal(240, 120, n_sessions)
                       ).round(0).astype(int),
    "device"         : np.random.choice(
                           ["Android", "iOS"], n_sessions,
                           p=[0.63, 0.37]),
    "pages_viewed"   : np.random.randint(1, 12, n_sessions)
})

# ── TABLE 3: events ──────────────────────────────────────────
# Granular action-level telemetry logs mapping specific feature interactions
event_types = [
    "view_dashboard", "view_paystub",
    "download_paystub", "transfer_salary",
    "view_reports", "account_settings",
    "invite_friend", "upgrade_premium"
]
event_weights = [0.25, 0.20, 0.15, 0.18,
                 0.10, 0.07, 0.03, 0.02]

n_events = 25000
events_data = pd.DataFrame({
    "event_id"    : [f"EVT{i:07d}" for i in range(1, n_events+1)],
    "user_id"     : np.random.choice(
                        users_data["user_id"], n_events),
    "event_name"  : np.random.choice(
                        event_types, n_events,
                        p=event_weights),
    "event_date"  : pd.date_range(
                        "2024-01-01", "2024-03-31",
                        periods=n_events
                    ).strftime("%Y-%m-%d"),
    "platform"    : np.random.choice(
                        ["Android", "iOS"], n_events,
                        p=[0.63, 0.37]),
    "feature"     : np.random.choice(
                        ["paystub", "transfer",
                         "reports", "onboarding", "other"],
                        n_events,
                        p=[0.30, 0.25, 0.20, 0.15, 0.10])
})

# ── TABLE 4: transactions ────────────────────────────────────
# Financial activity logs tracking monetary flows and payment statuses
n_trx = 3000
transactions_data = pd.DataFrame({
    "trx_id"         : [f"TRX{i:06d}" for i in range(1, n_trx+1)],
    "user_id"        : np.random.choice(
                           users_data["user_id"], n_trx),
    "trx_date"       : pd.date_range(
                           "2024-01-01", "2024-03-31",
                           periods=n_trx
                       ).strftime("%Y-%m-%d"),
    "trx_type"       : np.random.choice(
                           ["transfer_salary", "cash_withdrawal",
                            "bill_payment"],
                           n_trx, p=[0.55, 0.25, 0.20]),
    "amount_millions": np.abs(
                           np.random.normal(5.2, 3.1, n_trx)
                       ).round(2),
    "status"         : np.random.choice(
                           ["success", "failed", "pending"],
                           n_trx, p=[0.88, 0.08, 0.04])
})

# ── Load Tables into In-Memory SQLite Engine ─────────────────
users_data.to_sql("users", conn, if_exists="replace", index=False)
sessions_data.to_sql("sessions", conn, if_exists="replace", index=False)
events_data.to_sql("events", conn, if_exists="replace", index=False)
transactions_data.to_sql("transactions", conn, if_exists="replace", index=False)

print("✅ GajiKu Analytics Database Successfully Initialized!")
print(f"\n   4 Relational Tables Active:")
print(f"   • users        : {len(users_data):,} rows")
print(f"   • sessions     : {len(sessions_data):,} rows")
print(f"   • events       : {len(events_data):,} rows")
print(f"   • transactions : {len(transactions_data):,} rows")
print(f"""
   Database Schema:
   users(user_id, name, user_type, city, device, channel, registration_date, plan)
   sessions(session_id, user_id, session_date, duration_sec, device, pages_viewed)
   events(event_id, user_id, event_name, event_date, platform, feature)
   transactions(trx_id, user_id, trx_date, trx_type, amount_millions, status)
""")
````

````python
# ============================================================
# CELL 3 — Mental Framework: SPSS / GUI Analytics → SQL Translation Matrix
# Keep this reference card handy to rapidly translate statistical
# operations into relational database query logic!
# ============================================================

print("""
╔══════════════════════════════════════════════════════════════════════════════════╗
║                   COMPREHENSIVE TRANSLATION MATRIX: SPSS → SQL                   ║
╠══════════════════════════════════════════════════════════════════════════════════╣
║                                                                                  ║
║  ANALYTICAL OPERATION    SPSS GUI ACTION          SQL SYNTAX                     ║
║  ──────────────────────────────────────────────────────────────────────────────  ║
║  Select Variables        Click column header      SELECT column_name             ║
║  Retrieve All Variables  Active Data View         SELECT *                       ║
║  Identify Source Table   Active Data File (.sav)  FROM table_name                ║
║  Filter Rows / Cases     Data → Select Cases      WHERE condition                ║
║  Sort Observations       Data → Sort Cases        ORDER BY column_name           ║
║  Limit Row Output        (No direct equivalent)   LIMIT n                        ║
║  Transform / Compute     Transform → Compute      column AS alias                ║
║  Group & Aggregate       Analyze → Descriptive    GROUP BY + Aggregate Function  ║
║                          Statistics → Explore                                    ║
║  Calculate Mean          Mean in Descriptives     AVG(column_name)               ║
║  Count Frequency         Count in Frequencies     COUNT(*) or COUNT(id)          ║
║  Calculate Sum           Sum in Descriptives      SUM(column_name)               ║
║  Find Max / Min          Max/Min in Descriptives  MAX(column) / MIN(column)      ║
║  Merge Datasets          Data → Merge Files       JOIN ... ON ...                ║
║  Remove Duplicates       Data → Identify/Remove   DISTINCT                       ║
║                          Duplicate Cases                                         ║
║  Filter Post-Aggregation (No direct equivalent)   HAVING condition               ║
║                                                                                  ║
║  CORE STRUCTURAL ANALOGY:                                                        ║
║  SELECT   = "Which specific variables do you want to inspect?"                   ║
║  FROM     = "Which dataset file or table holds this data?"                       ║
║  WHERE    = "Which individual rows/cases meet your filter criteria?"             ║
║  GROUP BY = "Which categorical variable are you using to 'Split File'?"          ║
║  ORDER BY = "How should the final output be sorted?"                             ║
║  JOIN     = "How are two relational files merged using a key ID?"                ║
║                                                                                  ║
╚══════════════════════════════════════════════════════════════════════════════════╝
""")
````

````python
# ============================================================
# CELL 4 — Part 1: SELECT & FROM Fundamental Queries
# Objective: Extract raw columns and calculate derived variables.
# SQL Analogy: Selecting specific variables from a dataset view.
# ============================================================

print("=" * 60)
print("     PART 1: SELECT & FROM — FOUNDATIONAL QUERIES")
print("=" * 60)

# ── Query 1: Retrieve All Columns ───────────────────────────
print("\n📌 Query 1: Preview all user record attributes (First 5 Rows)")
print("  SQL Syntax:")
q1 = """
SELECT *
FROM   users
LIMIT  5;
"""
print(q1)
sql(q1)

# ── Query 2: Select Specific Columns ────────────────────────
print("\n📌 Query 2: Select specific attributes only")
print("  (Equivalent to selecting specific variables in SPSS Data View)")
q2 = """
SELECT user_id,
       user_type,
       city,
       plan
FROM   users
LIMIT  5;
"""
print(q2)
sql(q2)

# ── Query 3: Computed Derived Columns ───────────────────────
print("\n📌 Query 3: Compute transformed duration and engagement metrics")
print("  (Equivalent to Transform → Compute Variable in SPSS)")
q3 = """
SELECT session_id,
       user_id,
       duration_sec,
       ROUND(duration_sec / 60.0, 2)            AS duration_minutes,
       pages_viewed,
       ROUND(duration_sec * 1.0 / pages_viewed, 1) AS sec_per_page
FROM   sessions
LIMIT  8;
"""
print(q3)
sql(q3)
````

````python
# ============================================================
# CELL 5 — Part 2: WHERE Clause — Case Selection & Data Filtering
# Objective: Isolate target observations based on conditional filters.
# SQL Analogy: Equivalent to Data → Select Cases in SPSS.
# ============================================================

print("=" * 60)
print("     PART 2: WHERE — CASE SELECTION & DATA FILTERING")
print("=" * 60)

# ── Query 4: Single-Condition Filtering ───────────────────────
print("\n📌 Query 4: Filter for Premium Tier subscribers only")
print("  (SPSS Analogy: Data → Select Cases → If condition: plan='premium')")
q4 = """
SELECT user_id,
       user_type,
       city,
       plan
FROM   users
WHERE  plan = 'premium'
LIMIT  8;
"""
print(q4)
result4 = sql(q4)
total_premium = sql(
    "SELECT COUNT(*) as n FROM users WHERE plan='premium'",
    show_rows=0
)["n"][0]
print(f"\n  Total Premium Subscribers: {total_premium:,}")

# ── Query 5: Logical AND Intersection Filter ─────────────────
print("\n📌 Query 5: Premium Subscribers located in Jakarta")
print("  (SPSS Analogy: Select Cases → plan='premium' AND city='Jakarta')")
q5 = """
SELECT user_id,
       user_type,
       city,
       channel,
       plan
FROM   users
WHERE  plan = 'premium'
  AND  city = 'Jakarta'
LIMIT  8;
"""
print(q5)
sql(q5)

# ── Query 6: Logical OR Inclusive Filter ─────────────────────
print("\n📌 Query 6: Users located in either Bandung OR Surabaya")
q6 = """
SELECT user_id,
       city,
       user_type,
       plan
FROM   users
WHERE  city = 'Bandung'
    OR city = 'Surabaya'
ORDER BY city
LIMIT  10;
"""
print(q6)
sql(q6)

# ── Query 7: Streamlined Membership Filter with IN ───────────
print("\n📌 Query 7: Clean set membership filtering using IN operator")
print("  (Cleaner, production-grade equivalent of Query 6)")
q7 = """
SELECT user_id,
       city,
       user_type,
       plan
FROM   users
WHERE  city IN ('Bandung', 'Surabaya')
ORDER BY city
LIMIT  10;
"""
print(q7)
sql(q7)

# ── Query 8: Numeric Range Filtering with BETWEEN ───────────
print("\n📌 Query 8: Sessions with duration between 2 and 5 minutes (120–300s)")
q8 = """
SELECT session_id,
       user_id,
       duration_sec,
       ROUND(duration_sec / 60.0, 1) AS duration_minutes
FROM   sessions
WHERE  duration_sec BETWEEN 120 AND 300
ORDER BY duration_sec DESC
LIMIT  8;
"""
print(q8)
sql(q8)

# ── Query 9: Substring Pattern Matching with LIKE ─────────────
print("\n📌 Query 9: Distinct events involving paystub interactions")
q9 = """
SELECT DISTINCT event_name
FROM   events
WHERE  event_name LIKE '%paystub%';
"""
print(q9)
sql(q9)
````

````python
# ============================================================
# CELL 6 — Part 3: GROUP BY & Aggregation Functions
# Objective: Categorize observations and calculate summary statistics.
# SQL Analogy: Equivalent to Analyze → Descriptive Statistics / Split File in SPSS.
# ============================================================

print("=" * 60)
print("     PART 3: GROUP BY — CATEGORICAL AGGREGATION & SUMMARY STATS")
print("=" * 60)

# ── Query 10: Frequency Distribution & Conditional Aggregation ──
print("\n📌 Query 10: Total user volume and premium penetration rates by city")
print("  (SPSS Analogy: Analyze → Descriptives → Split File by City → Count)")
q10 = """
SELECT city,
       COUNT(*)              AS total_users,
       COUNT(CASE WHEN plan = 'premium'
                  THEN 1 END) AS premium_users,
       ROUND(
           COUNT(CASE WHEN plan = 'premium'
                      THEN 1 END) * 100.0 / COUNT(*),
           1
       )                     AS pct_premium
FROM   users
GROUP  BY city
ORDER  BY total_users DESC;
"""
print(q10)
sql(q10)

# ── Query 11: Grouped Summary Statistics (AVG, MIN, MAX) ────────
print("\n📌 Query 11: Average session duration and engagement depth by device type")
print("  (SPSS Analogy: Analyze → Compare Means → Group by Device)")
q11 = """
SELECT device,
       COUNT(*)                              AS total_sessions,
       ROUND(AVG(duration_sec), 1)           AS avg_duration_sec,
       ROUND(AVG(duration_sec) / 60.0, 2)   AS avg_duration_min,
       ROUND(AVG(pages_viewed), 2)          AS avg_pages_viewed,
       MIN(duration_sec)                     AS min_duration_sec,
       MAX(duration_sec)                     AS max_duration_sec
FROM   sessions
GROUP  BY device
ORDER  BY avg_duration_sec DESC;
"""
print(q11)
sql(q11)

# ── Query 12: Multi-Factor Segmented Grouping ─────────────────
print("\n📌 Query 12: Session duration broken down by Device × User Type")
print("  (SPSS Analogy: Split File by Device & User Type)")
q12 = """
SELECT s.device,
       u.user_type,
       COUNT(*)                             AS total_sessions,
       ROUND(AVG(s.duration_sec) / 60.0,
             2)                             AS avg_duration_min,
       ROUND(AVG(s.pages_viewed), 2)        AS avg_pages_viewed
FROM   sessions s
JOIN   users    u ON s.user_id = u.user_id
GROUP  BY s.device,
          u.user_type
ORDER  BY s.device,
          avg_duration_min DESC;
"""
print(q12)
sql(q12)
````

````python
# ============================================================
# CELL 7 — Part 4: Relational JOINs — Multi-Table Merges
# Objective: Merge primary and foreign key attributes across discrete tables.
# SQL Analogy: Equivalent to Data → Merge Files (Key Matching) in SPSS.
# ============================================================

print("""
╔══════════════════════════════════════════════════════════════════════════════════╗
║                       RELATIONAL JOIN TYPES MATRIX                              ║
╠══════════════════════════════════════════════════════════════════════════════════╣
║                                                                                  ║
║  INNER JOIN  → Returns ONLY observations with matching keys in BOTH tables.      ║
║               (SPSS Analogy: Merge Files retaining matched cases only).          ║
║               users: 500 | sessions: 8,000 → Retains records in both.           ║
║                                                                                  ║
║  LEFT JOIN   → Returns ALL records from Left table + matching Right table rows   ║
║               (Fills NULL values where right-table matches are absent).          ║
║               Example: Retains all users, including those with zero sessions.    ║
║                                                                                  ║
║  PRACTICAL ANALOGY:                                                              ║
║  INNER JOIN = Respondents who fully completed ALL survey blocks.                 ║
║  LEFT JOIN  = Entire sample population, including non-respondents/partial skips. ║
║                                                                                  ║
╚══════════════════════════════════════════════════════════════════════════════════╝
""")

# ── Query 13: Foundational INNER JOIN ─────────────────────────
print("📌 Query 13: Merge session telemetry with master user profiles")
print("  (Join Primary/Foreign Key: user_id)")
q13 = """
SELECT s.session_id,
       s.user_id,
       u.user_type,
       u.city,
       u.plan,
       s.session_date,
       s.duration_sec,
       s.pages_viewed
FROM   sessions s
JOIN   users    u ON s.user_id = u.user_id
LIMIT  8;
"""
print(q13)
sql(q13)

# ── Query 14: Cross-Table Engagement Aggregation ─────────────
print("\n📌 Query 14: User engagement profiles segmented by user type and plan tier")
print("  (Cross-functional diagnostic query for Product Leadership)")
q14 = """
SELECT u.user_type,
       u.plan,
       COUNT(DISTINCT u.user_id)           AS total_users,
       COUNT(DISTINCT s.session_id)        AS total_sessions,
       ROUND(
           COUNT(DISTINCT s.session_id) * 1.0 /
           COUNT(DISTINCT u.user_id),
           2
       )                                   AS sessions_per_user,
       ROUND(AVG(s.duration_sec) / 60.0,
             2)                            AS avg_duration_min
FROM   users    u
LEFT JOIN sessions s ON u.user_id = s.user_id
GROUP  BY u.user_type,
          u.plan
ORDER  BY sessions_per_user DESC;
"""
print(q14)
sql(q14)

# ── Query 15: Multi-Table Relational Triangulation ───────────
print("\n📌 Query 15: Triangulate user profiles across behavioral events and transactions")
print("  (Consolidating users + events + transactions tables)")
q15 = """
SELECT u.user_id,
       u.user_type,
       u.plan,
       COUNT(DISTINCT e.event_id)  AS total_events,
       COUNT(DISTINCT t.trx_id)    AS total_transactions,
       ROUND(SUM(
           CASE WHEN t.status = 'success'
                THEN t.amount_millions ELSE 0
           END
       ), 2)                       AS total_successful_value_mil
FROM   users        u
LEFT JOIN events    e ON u.user_id = e.user_id
LEFT JOIN transactions t ON u.user_id = t.user_id
GROUP  BY u.user_id,
          u.user_type,
          u.plan
HAVING total_events > 0
ORDER  BY total_transactions DESC
LIMIT  10;
"""
print(q15)
sql(q15)
````

````python
# ============================================================
# CELL 8 — Part 5: HAVING & ORDER BY — Post-Aggregation Filtering
# Objective: Apply conditional thresholds to aggregated group statistics.
# SQL Analogy: Filtering summary table outputs post-descriptive calculations.
# ============================================================

print("=" * 60)
print("     PART 5: HAVING & ORDER BY — POST-AGGREGATION FILTERING")
print("=" * 60)

print("""
  Key Distinction: WHERE vs. HAVING
  ──────────────────────────────────────────────────────────────────────────
  WHERE   → Filters individual records BEFORE grouping/aggregation.
  HAVING  → Filters grouped summary metrics AFTER aggregation occurs.

  SPSS Analogy:
  WHERE  = Data → Select Cases (executed prior to running procedure).
  HAVING = Filtering output table rows based on computed summary statistics.
  ──────────────────────────────────────────────────────────────────────────
""")

# ── Query 16: Aggregated Threshold Filtering with HAVING ──────
print("📌 Query 16: Identify Power Users generating > 20 total sessions")
print("  (Note: WHERE cannot filter on aggregated metrics like COUNT or AVG!)")
q16 = """
SELECT u.user_id,
       u.user_type,
       u.city,
       u.plan,
       COUNT(s.session_id) AS total_sessions
FROM   users    u
JOIN   sessions s ON u.user_id = s.user_id
GROUP  BY u.user_id,
          u.user_type,
          u.city,
          u.plan
HAVING total_sessions > 20
ORDER  BY total_sessions DESC
LIMIT  10;
"""
print(q16)
result16 = sql(q16)
print(f"\n  Total Power Users (> 20 Sessions): {len(result16):,}")

# ── Query 17: Combining WHERE (Pre-Filter) & HAVING (Post-Filter) ──
print("\n📌 Query 17: Premium Jakarta users generating > 15 total sessions")
q17 = """
SELECT u.user_id,
       u.user_type,
       u.city,
       COUNT(s.session_id)                 AS total_sessions,
       ROUND(AVG(s.duration_sec) / 60.0,
             2)                            AS avg_duration_min
FROM   users    u
JOIN   sessions s ON u.user_id = s.user_id
WHERE  u.plan = 'premium'            -- Filter BEFORE grouping
  AND  u.city = 'Jakarta'
GROUP  BY u.user_id,
          u.user_type,
          u.city
HAVING total_sessions > 15           -- Filter AFTER grouping
ORDER  BY total_sessions DESC
LIMIT  8;
"""
print(q17)
sql(q17)
````

````python
# ============================================================
# CELL 9 — Part 6: CASE WHEN — Conditional Logic & Recoding
# Objective: Generate categorical buckets and conditional metrics.
# SQL Analogy: Equivalent to Transform → Recode into Different Variables in SPSS.
# ============================================================

print("=" * 60)
print("     PART 6: CASE WHEN — CONDITIONAL LOGIC & VARIABLE RECODING")
print("=" * 60)

print("""
  CASE WHEN = Transform → Recode into Different Variables (SPSS)
  Creates dynamic new columns or conditional metrics based on logical rules.
""")

# ── Query 18: Behavioral User Segmentation ───────────────────
print("📌 Query 18: Segment users into activity tiers based on session counts")
q18 = """
SELECT u.user_id,
       u.user_type,
       COUNT(s.session_id)   AS total_sessions,
       CASE
           WHEN COUNT(s.session_id) >= 20
               THEN 'Power User'
           WHEN COUNT(s.session_id) BETWEEN 10 AND 19
               THEN 'Regular User'
           WHEN COUNT(s.session_id) BETWEEN 1 AND 9
               THEN 'Casual User'
           ELSE 'Inactive'
       END                   AS engagement_segment
FROM   users    u
LEFT JOIN sessions s ON u.user_id = s.user_id
GROUP  BY u.user_id, u.user_type
ORDER  BY total_sessions DESC
LIMIT  12;
"""
print(q18)
sql(q18)

# ── Query 19: Conditional Aggregation & Success Rates ─────────
print("\n📌 Query 19: Compute transaction success and failure ratios by transaction type")
q19 = """
SELECT trx_type,
       COUNT(*)                          AS total_trx,
       SUM(CASE WHEN status = 'success'
                THEN 1 ELSE 0 END)       AS successful_trx,
       SUM(CASE WHEN status = 'failed'
                THEN 1 ELSE 0 END)       AS failed_trx,
       ROUND(
           SUM(CASE WHEN status = 'success'
                    THEN 1.0 ELSE 0 END)
           / COUNT(*) * 100, 2
       )                                 AS success_rate_pct
FROM   transactions
GROUP  BY trx_type
ORDER  BY success_rate_pct DESC;
"""
print(q19)
sql(q19)
````

````python
# ============================================================
# CELL 10 — Part 7: Subqueries & Common Table Expressions (CTEs)
# Objective: Process complex multi-step data pipelines and modular queries.
# SQL Analogy: Staging temporary variables or intermediate dataset views.
# ============================================================

print("""
╔══════════════════════════════════════════════════════════════════════════════════╗
║                   SUBQUERIES vs. CTEs — WHEN TO USE WHICH?                       ║
╠══════════════════════════════════════════════════════════════════════════════════╣
║                                                                                  ║
║  SUBQUERY  = A query nested directly inside another query.                       ║
║              Best for: Simple inline scalar filters or quick lookup tables.      ║
║                                                                                  ║
║  CTE (WITH clause) = "Temporary Named Result Set"                                ║
║              Best for: Multi-step, complex analytical logic.                     ║
║              Significantly cleaner and more readable!                            ║
║                                                                                  ║
║  Analytical Analogy: CTEs function like creating named temporary dataset          ║
║  views in SPSS before executing the primary analysis block.                      ║
║                                                                                  ║
║  Industry Standard: Always prefer CTEs over nested subqueries for scripts > 10   ║
║  lines to ensure top-to-bottom logical clarity during code reviews.              ║
║                                                                                  ║
╚══════════════════════════════════════════════════════════════════════════════════╝
""")

# ── Query 20: Multi-Nested Subquery Approach ─────────────────
print("📌 Query 20: Retrieve users with above-average session counts")
print("  (Executed using traditional nested subqueries)")
q20 = """
SELECT u.user_id,
       u.user_type,
       u.plan,
       session_counts.total_sessions
FROM   users u
JOIN (
    -- Subquery 1: Calculate total sessions per user
    SELECT user_id,
           COUNT(*) AS total_sessions
    FROM   sessions
    GROUP  BY user_id
) AS session_counts ON u.user_id = session_counts.user_id
WHERE session_counts.total_sessions > (
    -- Subquery 2: Calculate baseline mean session count
    SELECT AVG(cnt)
    FROM (
        SELECT COUNT(*) AS cnt
        FROM   sessions
        GROUP  BY user_id
    )
)
ORDER BY session_counts.total_sessions DESC
LIMIT  10;
"""
print(q20)
sql(q20)

# ── Query 21: Modular CTE Approach (WITH Clause) ──────────────
print("\n📌 Query 21: Retrieve users with above-average session counts")
print("  (Executed using modular CTEs — Identical output, superior readability)")
q21 = """
-- CTE 1: Calculate total sessions per user
WITH sessions_per_user AS (
    SELECT user_id,
           COUNT(*) AS total_sessions
    FROM   sessions
    GROUP  BY user_id
),

-- CTE 2: Compute overall population mean
avg_sessions AS (
    SELECT AVG(total_sessions) AS population_avg
    FROM   sessions_per_user
)

-- Primary Execution: Join CTE outputs and filter above baseline
SELECT u.user_id,
       u.user_type,
       u.plan,
       spu.total_sessions,
       ROUND(ag.population_avg, 2) AS overall_avg_sessions,
       ROUND(
           spu.total_sessions - ag.population_avg,
           2
       )                           AS delta_from_avg
FROM   users             u
JOIN   sessions_per_user spu ON u.user_id = spu.user_id
CROSS JOIN avg_sessions  ag
WHERE  spu.total_sessions > ag.population_avg
ORDER  BY spu.total_sessions DESC
LIMIT  10;
"""
print(q21)
sql(q21)
print("\n  💡 CTEs dramatically enhance script maintainability!")
print("     Every WITH block serves as a clean, named analytical step.")
````

````python
# ============================================================
# CELL 2 — Concept: Funnel Analysis
# Objective: Understand multi-step conversion flows and identify drop-off points.
# Business Analogy: Quantifying attrition and drop-outs across longitudinal study phases.
# ============================================================

print("""
╔══════════════════════════════════════════════════════════════════════════════════╗
║                             FUNNEL ANALYSIS                                      ║
╠══════════════════════════════════════════════════════════════════════════════════╣
║                                                                                  ║
║  Definition: Measuring user completion velocity and drop-off volume across       ║
║  sequential steps in a defined product journey or user flow.                      ║
║                                                                                  ║
║  Example Journey: GajiKu Salary Transfer Flow                                    ║
║                                                                                  ║
║  [View Home] → [Select Transfer] → [Enter Amount] →                             ║
║  [Select Account] → [Confirm Details] → [Input PIN] → [Transfer Success]         ║
║                                                                                  ║
║  Key Metrics per Step:                                                           ║
║  • Users Reached    : Total unique users initiating/completing this step         ║
║  • Drop-off Volume  : Number of users exiting before reaching the step           ║
║  • Drop-off Rate (%) : Percentage of users lost relative to the previous step    ║
║  • Conversion Rate  : Cumulative % retained relative to top-of-funnel baseline   ║
║                                                                                  ║
║  Analytical Context:                                                             ║
║  • Funnel Analysis = Attrition analysis in longitudinal tracking studies         ║
║  • Step Drop-off    = Participant dropout rate at specific milestones            ║
║  • Research Goal    = Identify exact friction points causing user exit          ║
║                                                                                  ║
╚══════════════════════════════════════════════════════════════════════════════════╝
""")
````

````python
# ============================================================
# CELL 3 — Funnel Analysis: Complete SQL Pipeline
# Objective: Calculate step-by-step retention, absolute drop-off volume,
# step drop-off percentage, and overall conversion from baseline.
# ============================================================

q_funnel = """
-- ============================================================
-- FUNNEL ANALYSIS: GajiKu Salary Transfer Checkout Journey
-- Quantifying user progression and attrition across sequence steps
-- ============================================================

WITH

-- CTE 1: Count distinct users per journey milestone
step_counts AS (
    SELECT step_name,
           step_order,
           COUNT(DISTINCT user_id) AS users_reached
    FROM   funnel_events
    GROUP  BY step_name,
              step_order
),

-- CTE 2: Capture top-of-funnel user volume (Step 1 baseline)
baseline AS (
    SELECT users_reached AS total_start
    FROM   step_counts
    WHERE  step_order = 1
),

-- CTE 3: Compute previous step user counts via window functions
funnel_metrics AS (
    SELECT s.step_order,
           s.step_name,
           s.users_reached,
           LAG(s.users_reached) OVER (
               ORDER BY s.step_order
           )                       AS users_prev_step,
           b.total_start
    FROM   step_counts s
    CROSS  JOIN baseline b
)

-- Final Execution: Output comprehensive drop-off & conversion rates
SELECT step_order,
       step_name,
       users_reached,

       -- Absolute drop-off count relative to previous step
       COALESCE(users_prev_step - users_reached, 0)
                                    AS dropoff_count,

       -- Step drop-off rate (% lost vs. previous step)
       CASE WHEN users_prev_step IS NOT NULL
            THEN ROUND(
                (1.0 - users_reached * 1.0 / users_prev_step)
                * 100, 1)
            ELSE 0.0
       END                          AS dropoff_rate_pct,

       -- Cumulative conversion rate (% retained vs. baseline)
       ROUND(users_reached * 100.0 / total_start, 1)
                                    AS pct_from_start

FROM   funnel_metrics
ORDER  BY step_order;
"""

df_funnel = sql(q_funnel)

print(f"""
  Executive Summary Summary:
  ─────────────────────────────────────────────────────
  Top-of-Funnel Volume (Start) : {df_funnel['users_reached'].max():,}
  Bottom-of-Funnel Volume (End): {df_funnel['users_reached'].min():,}
  Overall Funnel Completion    : {df_funnel['pct_from_start'].min():.1f}%
  Highest Step Drop-off Friction: {df_funnel.loc[df_funnel['dropoff_rate_pct'].idxmax(), 'step_name']}
                                  ({df_funnel['dropoff_rate_pct'].max():.1f}%)
""")
````

````python
# ============================================================
# CELL 5 — Funnel Segmentation: Platform Breakdown
# PM Objective: "Are there significant conversion discrepancies
#                between Android and iOS user journeys?"
# ============================================================

q_funnel_device = """
-- ============================================================
-- Platform-specific funnel breakdown from events telemetry
-- ============================================================
WITH

platform_funnel AS (
    SELECT
        e.platform,
        e.event_name,
        COUNT(DISTINCT e.user_id) AS users_reached
    FROM   events e
    WHERE  e.event_name IN (
               'app_open',
               'view_slip',
               'download_slip',
               'transfer_init',
               'transfer_success'
           )
    GROUP  BY e.platform,
              e.event_name
),

-- Capture platform-specific baseline volume (Top-of-funnel)
baseline_per_platform AS (
    SELECT platform,
           users_reached AS total_start
    FROM   platform_funnel
    WHERE  event_name = 'app_open'
)

-- Execute cumulative conversion calculation per platform
SELECT
    pf.platform,
    pf.event_name,
    pf.users_reached,
    ROUND(
        pf.users_reached * 100.0 / bp.total_start,
        1
    )                        AS pct_from_start
FROM   platform_funnel       pf
JOIN   baseline_per_platform bp
       ON pf.platform = bp.platform
ORDER  BY pf.platform,
          pf.users_reached DESC;
"""

df_funnel_dev = sql(q_funnel_device)

# ── Pivot table for comparative visualization ─────────────────
df_pivot = df_funnel_dev.pivot(
    index="event_name",
    columns="platform",
    values="pct_from_start"
).fillna(0)

# Sort funnel stages sequentially mapping
step_order_map = {
    "app_open"         : 1,
    "view_slip"        : 2,
    "download_slip"    : 3,
    "transfer_init"    : 4,
    "transfer_success" : 5
}
df_pivot["order"] = df_pivot.index.map(
    lambda x: step_order_map.get(x, 99)
)
df_pivot = df_pivot.sort_values("order").drop("order", axis=1)

print("\n  Funnel Conversion (%) by Platform:")
print(df_pivot.round(1).to_string())

# Identify funnel stage with highest platform friction gap
if "Android" in df_pivot.columns and "iOS" in df_pivot.columns:
    df_pivot["gap"] = (
        df_pivot["iOS"] - df_pivot["Android"]
    ).abs()
    worst_step = df_pivot["gap"].idxmax()
    worst_gap  = df_pivot["gap"].max()
    print(f"\n  ⚠️  Largest Discrepancy Point: {worst_step}")
    print(f"      iOS vs Android Gap: {worst_gap:.1f}%")
    print("      → PM Action Item: Prioritize technical/UX investigation for this specific platform step.")
````

````python
# ============================================================
# CELL 6 — Core Concept: Retention Cohort Analysis
# Principle: "Retention is the ultimate product metric. Net Promoter
# Scores can be inflated, but if users do not return,
# the product fails to deliver sustainable value."
# ============================================================

print("""
╔══════════════════════════════════════════════════════════════════╗
║                COHORT RETENTION — CORE CONCEPTS                  ║
╠══════════════════════════════════════════════════════════════════╣
║                                                                  ║
║  COHORT    = A segment of users who performed their first action ║
║              (e.g., registered/onboarded) during the same period ║
║                                                                  ║
║  RETENTION = % of users in that cohort who remain active         ║
║              N days / weeks / months after acquisition           ║
║                                                                  ║
║  Sample Monthly Cohort Retention Matrix:                         ║
║  ─────────────────────────────────────────────────────────────   ║
║  Cohort    │ Month 0 │ Month 1 │ Month 2 │ Month 3               ║
║  ──────────┼─────────┼─────────┼─────────┼────────               ║
║  Jan 2024  │  100%   │   45%   │   32%   │   28%                 ║
║  Feb 2024  │  100%   │   51%   │   37%   │    -                  ║
║  Mar 2024  │  100%   │   48%   │    -    │    -                  ║
║                                                                  ║
║  How to Interpret:                                               ║
║  • Jan 2024 Cohort: Out of 100 onboarded users in January,       ║
║    45 remained active in February (45% M1 Retention).            ║
║  • Horizontal Drift: Tracks user decay over time.                ║
║  • Vertical Drift: Compares quality across acquisition cohorts.  ║
║                                                                  ║
║  Standard Retention Milestones:                                  ║
║  • D1 Retention  = % of users returning 1 day post-signup.       ║
║  • D7 Retention  = % of users returning within 7 days.           ║
║  • D30 Retention = % of users remaining active after 30 days.    ║
║                                                                  ║
╚══════════════════════════════════════════════════════════════════╝
""")
````

````python
# ============================================================
# CELL 7 — Synthetic Cohort Data Generation (Fixed Day 0 Baseline)
# ============================================================

np.random.seed(42)

cohort_rows = []

for _, user in users_df.iterrows():
    reg_col = "registration_date" if "registration_date" in user else "tanggal_daftar"
    reg_date = pd.to_datetime(user[reg_col])

    if reg_date > pd.to_datetime("2024-02-28"):
        continue

    base_retention = 0.55 if user["plan"] == "premium" else 0.35

    # Day 0 is ALWAYS recorded for baseline cohort size (100% onboarding)
    cohort_rows.append({
        "user_id"           : user["user_id"],
        "plan"              : user["plan"],
        "registration_date" : reg_date.strftime("%Y-%m-%d"),
        "cohort_month"      : reg_date.strftime("%Y-%m"),
        "activity_date"     : reg_date.strftime("%Y-%m-%d"),
        "days_since_reg"    : 0
    })

    # Simulate subsequent activity (Day 1 to Day 90)
    for day_offset in range(1, 91):
        p_active = base_retention * np.exp(-0.025 * day_offset) + np.random.normal(0, 0.03)
        p_active = max(0, min(1, p_active))

        if np.random.random() < p_active:
            active_date = reg_date + timedelta(days=day_offset)
            if active_date <= pd.to_datetime("2024-03-31"):
                cohort_rows.append({
                    "user_id"           : user["user_id"],
                    "plan"              : user["plan"],
                    "registration_date" : reg_date.strftime("%Y-%m-%d"),
                    "cohort_month"      : reg_date.strftime("%Y-%m"),
                    "activity_date"     : active_date.strftime("%Y-%m-%d"),
                    "days_since_reg"    : day_offset
                })

cohort_df = pd.DataFrame(cohort_rows)
cohort_df.to_sql("cohort_activity", conn, if_exists="replace", index=False)

print(f"✅ Cohort Activity Table Rebuilt: {len(cohort_df):,} records")
````

````python
# ============================================================
# CELL 8 — Milestone Cohort Retention Matrix (Exact Milestone Days)
# ============================================================

q_retention = """
WITH cohort_size AS (
    SELECT cohort_month,
           COUNT(DISTINCT user_id) AS total_users
    FROM   cohort_activity
    WHERE  days_since_reg = 0
    GROUP  BY cohort_month
),

retention_counts AS (
    SELECT
        ca.cohort_month,
        COUNT(DISTINCT CASE WHEN ca.days_since_reg = 1 THEN ca.user_id END)  AS active_d1,
        COUNT(DISTINCT CASE WHEN ca.days_since_reg = 3 THEN ca.user_id END)  AS active_d3,
        COUNT(DISTINCT CASE WHEN ca.days_since_reg = 7 THEN ca.user_id END)  AS active_d7,
        COUNT(DISTINCT CASE WHEN ca.days_since_reg = 14 THEN ca.user_id END) AS active_d14,
        COUNT(DISTINCT CASE WHEN ca.days_since_reg = 30 THEN ca.user_id END) AS active_d30,
        COUNT(DISTINCT CASE WHEN ca.days_since_reg = 60 THEN ca.user_id END) AS active_d60
    FROM   cohort_activity ca
    GROUP  BY ca.cohort_month
)

SELECT
    rc.cohort_month,
    cs.total_users                             AS cohort_size,
    ROUND(rc.active_d1  * 100.0 / cs.total_users, 1) AS d1_ret_pct,
    ROUND(rc.active_d3  * 100.0 / cs.total_users, 1) AS d3_ret_pct,
    ROUND(rc.active_d7  * 100.0 / cs.total_users, 1) AS d7_ret_pct,
    ROUND(rc.active_d14 * 100.0 / cs.total_users, 1) AS d14_ret_pct,
    ROUND(rc.active_d30 * 100.0 / cs.total_users, 1) AS d30_ret_pct,
    ROUND(rc.active_d60 * 100.0 / cs.total_users, 1) AS d60_ret_pct
FROM   retention_counts rc
JOIN   cohort_size       cs ON rc.cohort_month = cs.cohort_month
ORDER  BY rc.cohort_month;
"""

df_retention = sql(q_retention)
````

````python
# ============================================================
# CELL 11 — Strategic Framework: RFM User Segmentation
# Executive Overview: RFM is more than simple clustering — it provides
# a holistic health view of your entire user base within a single model.
# ============================================================

print("""
╔══════════════════════════════════════════════════════════════════════════════╗
║                          RFM SEGMENTATION FRAMEWORK                          ║
╠══════════════════════════════════════════════════════════════════════════════╣
║                                                                              ║
║  R = RECENCY   : Days elapsed since the user's most recent activity.         ║
║                  Lower score = Recently Active = HIGH VALUE                  ║
║                                                                              ║
║  F = FREQUENCY : Total active sessions/events logged within the observation  ║
║                  window.                                                     ║
║                  Higher score = Frequent Usage = HIGH VALUE                  ║
║                                                                              ║
║  M = MONETARY  : Cumulative financial contribution or transaction volume.     ║
║                  Higher score = High Lifetime Value (LTV) = HIGH VALUE       ║
║                                                                              ║
║  Scoring Methodology: Each dimension is ranked 1–4 using Quartiles NTILE(4)  ║
║  Total Composite Score = R + F + M (Ranging from 3 to 12)                    ║
║                                                                              ║
║  Strategic User Segments:                                                    ║
║  10–12 = Champions      → VIP Perks, Loyalty Programs & Early Access          ║
║  7–9   = Loyal Users    → Upsell Cross-Sells & Referral Incentives           ║
║  5–6   = At Risk        → Targeted Re-engagement & Retention Campaigns       ║
║  3–4   = Churned/Lost   → Win-Back Offers or Sunset Workflows                ║
║                                                                              ║
╚══════════════════════════════════════════════════════════════════════════════╝
""")
````

````python
# ============================================================
# CELL 12 — Comprehensive RFM Calculation Query (NTILE Quartiles)
# Benchmark Reference Date: 2024-03-31 (Observation Window End)
# ============================================================

q_rfm = """
-- ============================================================
-- RFM USER SEGMENTATION — GajiKu Active User Base
-- ============================================================

WITH

-- Step 1: Calculate raw R, F, and M metrics per user
rfm_raw AS (
    SELECT
        u.user_id,
        u.plan,
        u.city,
        u.channel,

        -- RECENCY: Days elapsed since most recent active session
        -- Lower values indicate recent activity (Higher LTV Potential)
        JULIANDAY('2024-03-31') -
        JULIANDAY(MAX(s.session_date)) AS recency_days,

        -- FREQUENCY: Total distinct active sessions logged (90-day window)
        COUNT(DISTINCT s.session_id)    AS frequency,

        -- MONETARY: Total successful transaction value (in IDR Millions)
        COALESCE(SUM(
            CASE WHEN LOWER(t.status) IN ('success', 'sukses')
                 THEN t.amount_millions
                 ELSE 0 END
        ), 0)                           AS monetary

    FROM   users        u
    LEFT JOIN sessions  s ON u.user_id = s.user_id
    LEFT JOIN transactions t ON u.user_id = t.user_id
    GROUP BY u.user_id, u.plan, u.city, u.channel
),

-- Step 2: Assign 1–4 percentile scores per dimension using NTILE
-- NTILE(4) divides users into 4 equal volume quartiles
rfm_scored AS (
    SELECT *,

        -- R Score: Lower recency_days = better performance → DESCENDING order
        NTILE(4) OVER (
            ORDER BY recency_days DESC
        )                   AS r_score,

        -- F Score: Higher session frequency = better performance → ASCENDING order
        NTILE(4) OVER (
            ORDER BY frequency ASC
        )                   AS f_score,

        -- M Score: Higher monetary volume = better performance → ASCENDING order
        NTILE(4) OVER (
            ORDER BY monetary ASC
        )                   AS m_score

    FROM rfm_raw
),

-- Step 3: Composite scoring & strategic user segment mapping
rfm_segments AS (
    SELECT *,
           r_score + f_score + m_score AS rfm_total,
           CASE
               WHEN r_score + f_score + m_score >= 10
                   THEN 'Champions'
               WHEN r_score + f_score + m_score >= 7
                   THEN 'Loyal Users'
               WHEN r_score + f_score + m_score >= 5
                   THEN 'At Risk'
               ELSE 'Lost / Churned'
           END                         AS rfm_segment
    FROM rfm_scored
)

SELECT
    user_id,
    plan,
    city,
    channel,
    ROUND(recency_days, 0) AS recency_days,
    frequency,
    ROUND(monetary, 2)     AS monetary_juta,
    r_score,
    f_score,
    m_score,
    rfm_total,
    rfm_segment
FROM rfm_segments
ORDER BY rfm_total DESC;
"""

df_rfm = sql(q_rfm, show_rows=12)

print(f"\n  RFM User Segment Distribution:")
seg_dist = df_rfm["rfm_segment"].value_counts()
for seg, n in seg_dist.items():
    pct = n / len(df_rfm) * 100
    bar = "█" * int(pct / 3)
    print(f"  {seg:<16}: {n:4d} ({pct:5.1f}%) {bar}")
````

````python
# ============================================================
# CELL 14 — Event Popularity & Feature Adoption Analysis
# Objective: Measure top-of-funnel engagement by tracking event volume,
# unique user reach, and feature adoption percentages across the platform.
# ============================================================

print("=" * 65)
print("     PATTERN 4: EVENT & SESSION PATH ANALYSIS")
print("=" * 65)

# ── 1. Event Popularity & Unique User Reach ─────────────────
q_event_pop = """
SELECT
    event_name,
    COUNT(*)                              AS total_occurrences,
    COUNT(DISTINCT user_id)               AS unique_users,
    ROUND(
        COUNT(DISTINCT user_id) * 100.0 /
        (SELECT COUNT(DISTINCT user_id)
         FROM events),
        1
    )                                     AS user_adoption_pct
FROM   events
GROUP  BY event_name
ORDER  BY unique_users DESC;
"""

print("\n  EVENT POPULARITY & ADOPTION METRICS:")
df_events_pop = sql(q_event_pop)

# ── 2. Core Feature Adoption Rates ───────────────────────────
q_adoption = """
-- Calculates the adoption rate (%) across key key product interactions
WITH total_users AS (
    SELECT COUNT(DISTINCT user_id) AS total_count
    FROM   users
)
SELECT
    event_name                           AS feature_name,
    COUNT(DISTINCT e.user_id)           AS active_users,
    ROUND(
        COUNT(DISTINCT e.user_id) * 100.0 / tu.total_count,
        1
    )                                   AS adoption_rate_pct
FROM   events e
CROSS  JOIN total_users tu
WHERE  event_name IN (
           'view_slip',
           'download_slip',
           'transfer_init',
           'transfer_success',
           'invite_friend',
           'upgrade_click'
       )
GROUP  BY event_name, tu.total_count
ORDER  BY adoption_rate_pct DESC;
"""

print("\n  CORE FEATURE ADOPTION RATES:")
df_adoption = sql(q_adoption)
````

````python
# ============================================================
# CELL 15 — Session Depth & Dead-End Event Analysis
# Objective: Evaluate user engagement depth per session and identify
# high-frequency exit events to diagnose funnel drop-offs.
# ============================================================

# ── 1. Session Depth Breakdown ──────────────────────────────
q_session_depth = """
-- Analyzes session depth (pages viewed per session) segmented by user type
WITH session_depth AS (
    SELECT
        s.user_id,
        s.session_id,
        s.pages_viewed,
        u.user_type,
        u.plan,
        CASE
            WHEN s.pages_viewed >= 8 THEN 'Deep (8+ pages)'
            WHEN s.pages_viewed >= 4 THEN 'Medium (4–7 pages)'
            WHEN s.pages_viewed >= 2 THEN 'Shallow (2–3 pages)'
            ELSE 'Bounce (1 page)'
        END                          AS session_type
    FROM   sessions s
    JOIN   users    u ON s.user_id = u.user_id
)
SELECT
    user_type,
    session_type,
    COUNT(*)                         AS total_sessions,
    ROUND(
        COUNT(*) * 100.0 /
        SUM(COUNT(*)) OVER (PARTITION BY user_type),
        1
    )                                AS pct_of_user_type
FROM   session_depth
GROUP  BY user_type, session_type
ORDER  BY user_type, total_sessions DESC;
"""

print("\n  SESSION DEPTH ANALYSIS:")
df_session_depth = sql(q_session_depth)

# ── 2. Dead-End Event Detection ─────────────────────────────
q_dead_end = """
-- Identifies the most common FINAL event prior to session exit
WITH user_last_event AS (
    SELECT
        user_id,
        event_name,
        ROW_NUMBER() OVER (
            PARTITION BY user_id
            ORDER BY event_id DESC
        ) AS event_rank
    FROM events
)
SELECT
    event_name                       AS last_event_before_exit,
    COUNT(*)                         AS frequency,
    ROUND(
        COUNT(*) * 100.0 /
        (SELECT COUNT(DISTINCT user_id) FROM events),
        1
    )                                AS pct_users
FROM   user_last_event
WHERE  event_rank = 1
GROUP  BY event_name
ORDER  BY frequency DESC;
"""

print("\n  DEAD-END DETECTION (Final Action Before App Exit):")
df_dead_end = sql(q_dead_end)

print(f"""
  Strategic Executive Interpretation:
  Highest frequency exit event = Primary churn/exit point
  → Option A: Natural task completion point (User finished intent)
  → Option B: Product friction point (User encountered barrier/frustration)

  Primary Exit Event: {df_dead_end.iloc[0]['last_event_before_exit']}
  → Action Item: Differentiate between natural exits vs. unexpected drop-offs via session recordings.
""")
````

````python
# ============================================================
# CELL 2 — Window Functions Conceptual Foundations
# Objective: Print architectural reference map for advanced SQL
# window functions (Ranking, Navigation, and Aggregations).
# ============================================================

print("""
╔══════════════════════════════════════════════════════════════════╗
║                   WINDOW FUNCTIONS — CONCEPT MAP                 ║
╠══════════════════════════════════════════════════════════════════╣
║                                                                  ║
║  GROUP BY  → Compresses rows   → Loses row-level granularity     ║
║  WINDOW FN → Appends calculated → Preserves ALL original rows    ║
║              columns                                             ║
║                                                                  ║
║  Core Syntax Pattern:                                            ║
║  function_name() OVER (                                          ║
║      PARTITION BY grouping_column   ← Defines window scope       ║
║      ORDER BY sequence_column       ← Defines order within window║
║      ROWS/RANGE frame_specification  ← Optional window frame     ║
║  )                                                               ║
║                                                                  ║
║  Key Analytic Functions Covered:                                 ║
║  ─────────────────────────────────────────────────────────────  ║
║  • ROW_NUMBER() : Unique sequential integer per partition       ║
║  • RANK()       : Rank order with gaps for ties                  ║
║  • DENSE_RANK() : Rank order WITHOUT gaps for ties               ║
║  • LAG(col, n)  : Value from N rows PREVIOUS                     ║
║  • LEAD(col, n) : Value from N rows FOLLOWING                    ║
║  • SUM() OVER   : Cumulative running totals                      ║
║  • AVG() OVER   : Moving/rolling averages                          ║
║  • NTILE(n)     : Bucket distribution (Quartiles/Percentiles)    ║
║                                                                  ║
╚══════════════════════════════════════════════════════════════════╝
""")
````

````python
# ============================================================
# CELL 3 — Ranking Window Functions (ROW_NUMBER, RANK, DENSE_RANK)
# Objective: Identify top users per city based on successful transaction
# volume and illustrate ranking evaluation behavior.
# ============================================================

print("=" * 60)
print("     RANKING WINDOW FUNCTIONS (ROW_NUMBER, RANK, DENSE_RANK)")
print("=" * 60)

q_ranking = """
-- ============================================================
-- TOP USERS PER CITY BY SUCCESSFUL TRANSACTION VOLUME
-- Demonstrates evaluation mechanics across ROW_NUMBER, RANK, and DENSE_RANK
-- ============================================================

WITH

-- Step 1: Calculate total successful transactions and volume per user
user_trx_count AS (
    SELECT
        t.user_id,
        u.city,
        u.plan,
        COUNT(*)               AS total_trx,
        SUM(t.amount_millions) AS total_volume_millions
    FROM transactions AS t
    INNER JOIN users  AS u
        ON t.user_id = u.user_id
    WHERE t.status = 'success'
    GROUP BY t.user_id, u.city, u.plan
),

-- Step 2: Apply window ranking methods partitioned by city
ranked_users AS (
    SELECT
        user_id,
        city,
        plan,
        total_trx,
        ROUND(total_volume_millions, 2) AS total_volume_millions,

        -- ROW_NUMBER: Strictly unique index per partition (no tie handling)
        ROW_NUMBER() OVER (
            PARTITION BY city
            ORDER BY total_trx DESC
        ) AS row_num,

        -- RANK: Handles ties with rank gaps (e.g., 1, 1, 3)
        RANK() OVER (
            PARTITION BY city
            ORDER BY total_trx DESC
        ) AS rank_with_gap,

        -- DENSE_RANK: Handles ties WITHOUT rank gaps (e.g., 1, 1, 2)
        DENSE_RANK() OVER (
            PARTITION BY city
            ORDER BY total_trx DESC
        ) AS dense_rank

    FROM user_trx_count
)

-- FINAL OUTPUT: Retrieve Top 3 users per city
SELECT
    city,
    row_num,
    user_id,
    plan,
    total_trx,
    total_volume_millions,
    rank_with_gap,
    dense_rank
FROM ranked_users
WHERE row_num <= 3
ORDER BY city ASC, row_num ASC;
"""

print("\n  Top 3 Users per City:")
df_ranking = sql(q_ranking, show_rows=12)

print("""
  Methodology Comparison:
  ─────────────────────────────────────────────────────
  • ROW_NUMBER  → 1, 2, 3, 4, 5 (Unique row ID, ideal for top-N filters)
  • RANK        → 1, 2, 2, 4, 5 (Gaps present after tied positions)
  • DENSE_RANK  → 1, 2, 2, 3, 4 (Consecutive integers, ideal for leaderboards)

  Best Practice: Use ROW_NUMBER() when filtering strict N records per group.
                 Use DENSE_RANK() when equal scores must receive equal rank.
""")
````

````python
# ============================================================
# CELL 4 — Navigational Window Functions (LAG): Month-over-Month Growth
# Objective: Calculate overall platform MoM trajectory across Revenue,
# Total Transactions, and Active Users with automated health status.
# ============================================================

print("=" * 60)
print("     LAG & LEAD — Month-over-Month Executive Analysis")
print("=" * 60)

q_mom = """
-- ============================================================
-- MONTH-OVER-MONTH (MoM) PLATFORM AGGREGATION
-- Compares current month metrics against previous month using LAG(col, 1)
-- ============================================================

WITH

-- Step 1: Monthly metric aggregation
monthly_metrics AS (
    SELECT
        STRFTIME('%Y-%m', transaction_date) AS month,

        -- Revenue (successful transactions)
        SUM(CASE WHEN status = 'success'
                 THEN amount_millions ELSE 0 END)
                                            AS revenue_millions,

        -- Transaction counts
        COUNT(*)                            AS total_trx,
        SUM(CASE WHEN status = 'success'
                 THEN 1 ELSE 0 END)         AS successful_trx,

        -- Unique transacting users
        COUNT(DISTINCT user_id)             AS unique_users

    FROM transactions
    GROUP BY STRFTIME('%Y-%m', transaction_date)
),

-- Step 2: Retrieve previous month values using LAG
monthly_with_prev AS (
    SELECT
        month,
        revenue_millions,
        total_trx,
        successful_trx,
        unique_users,

        -- LAG: Retrieve preceding month values
        LAG(revenue_millions, 1) OVER (ORDER BY month ASC) AS prev_revenue,
        LAG(total_trx, 1)       OVER (ORDER BY month ASC) AS prev_total_trx,
        LAG(unique_users, 1)    OVER (ORDER BY month ASC) AS prev_unique_users

    FROM monthly_metrics
)

-- Step 3: Compute MoM growth percentages
SELECT
    month,

    ROUND(revenue_millions, 2) AS revenue_millions,
    ROUND(prev_revenue, 2)     AS prev_revenue_millions,
    ROUND(
        (revenue_millions - prev_revenue)
        * 100.0 / NULLIF(prev_revenue, 0),
        1
    )                          AS revenue_growth_pct,

    total_trx,
    prev_total_trx,
    ROUND(
        (total_trx - prev_total_trx)
        * 100.0 / NULLIF(prev_total_trx, 0),
        1
    )                          AS trx_growth_pct,

    unique_users,
    prev_unique_users,
    ROUND(
        (unique_users - prev_unique_users)
        * 100.0 / NULLIF(prev_unique_users, 0),
        1
    )                          AS user_growth_pct

FROM monthly_with_prev
ORDER BY month ASC;
"""

df_mom = sql(q_mom)

# Automated executive summary calculation
if len(df_mom) >= 2:
    last = df_mom.dropna().iloc[-1]
    print(f"""
  MoM SUMMARY — {last['month']}
  ─────────────────────────────────────────────────
  Revenue growth : {last['revenue_growth_pct']:+.1f}%
  Trx growth     : {last['trx_growth_pct']:+.1f}%
  User growth    : {last['user_growth_pct']:+.1f}%

  Traffic Light Health Status:
  Revenue     : {"🟢" if last['revenue_growth_pct'] > 0 else "🔴"}
  Transactions: {"🟢" if last['trx_growth_pct'] > 0 else "🔴"}
  Users       : {"🟢" if last['user_growth_pct'] > 0 else "🔴"}
    """)
````

````python
# ============================================================
# CELL 5 — Running Totals & Moving Averages (MA3 & MA7)
# Objective: Calculate daily cumulative revenue trajectory and smooth
# daily volatility using 3-day and 7-day moving averages.
# ============================================================

print("=" * 60)
print("     WINDOW FRAMES — Running Total & Rolling Moving Averages")
print("=" * 60)

q_running = """
-- ============================================================
-- CUMULATIVE REVENUE & ROLLING MOVING AVERAGES
-- Provides smoothed trend metrics for executive visual dashboards
-- ============================================================

WITH

-- Step 1: Daily revenue aggregation
daily_revenue AS (
    SELECT
        transaction_date,
        SUM(CASE WHEN status = 'success'
                 THEN amount_millions ELSE 0 END) AS daily_rev,
        COUNT(DISTINCT user_id)                   AS daily_users
    FROM transactions
    GROUP BY transaction_date
),

-- Step 2: Calculate window frames for cumulative sum and rolling averages
with_rolling AS (
    SELECT
        transaction_date,
        ROUND(daily_rev, 2) AS daily_revenue_millions,
        daily_users,

        -- Running total: Cumulative summation from initial date to current row
        ROUND(
            SUM(daily_rev) OVER (
                ORDER BY transaction_date ASC
                ROWS BETWEEN UNBOUNDED PRECEDING
                         AND CURRENT ROW
            ),
            2
        )                   AS cumulative_revenue,

        -- 7-Day Moving Average: Average across current row + 6 preceding days
        ROUND(
            AVG(daily_rev) OVER (
                ORDER BY transaction_date ASC
                ROWS BETWEEN 6 PRECEDING
                         AND CURRENT ROW
            ),
            2
        )                   AS ma7_revenue,

        -- 3-Day Moving Average: Average across current row + 2 preceding days
        ROUND(
            AVG(daily_rev) OVER (
                ORDER BY transaction_date ASC
                ROWS BETWEEN 2 PRECEDING
                         AND CURRENT ROW
            ),
            2
        )                   AS ma3_revenue

    FROM daily_revenue
)

-- FINAL OUTPUT: Chronological sequence
SELECT *
FROM with_rolling
ORDER BY transaction_date ASC;
"""

df_running = sql(q_running, show_rows=8)
print(f"\n  Total Records Processed: {len(df_running)}")
print(f"  Final Cumulative Revenue: "
      f"Rp {df_running['cumulative_revenue'].max():,.2f} Million")
````

````python
# ============================================================
# CELL 7 — Professional Multi-Layer CTE Query: User Health Score
# Objective: Calculate a multi-dimensional User Health Score (0–100)
# combining Recency, Engagement, and Transaction metrics to identify
# churn risks before attrition occurs.
# ============================================================

query_profesional = """
-- ============================================================
-- QUERY  : User Health Score Dashboard
-- AUTHOR : Product Analytics Team
-- DATE   : 2024-03-31
-- PURPOSE: Evaluates individual user health based on a weighted
--          combination of Recency, Engagement, and Financial Activity.
-- INPUTS : users, sessions, transactions, events
-- OUTPUT : 1 record per user with composite health score & segment
-- ============================================================

WITH

-- ─────────────────────────────────────────────────────────
-- CTE 1: Engagement Metrics (Sessions Data)
-- Metrics: Total session count, average duration, recency days
-- ─────────────────────────────────────────────────────────
session_metrics AS (
    SELECT
        user_id,
        COUNT(DISTINCT session_id)            AS total_sessions,
        ROUND(AVG(duration_seconds) / 60.0, 2) AS avg_duration_minutes,
        MAX(session_date)                      AS last_session_date,
        JULIANDAY('2024-03-31') -
        JULIANDAY(MAX(session_date))           AS days_since_active
    FROM sessions
    GROUP BY user_id
),

-- ─────────────────────────────────────────────────────────
-- CTE 2: Transaction Metrics
-- Metrics: Total transactions, successful transactions, volume, success rate
-- ─────────────────────────────────────────────────────────
transaction_metrics AS (
    SELECT
        user_id,
        COUNT(*)                              AS total_trx,
        SUM(CASE WHEN status = 'success'
                 THEN 1 ELSE 0 END)           AS successful_trx,
        ROUND(
            SUM(CASE WHEN status = 'success'
                     THEN amount_millions ELSE 0 END),
            2
        )                                     AS total_volume_millions,
        ROUND(
            SUM(CASE WHEN status = 'success'
                     THEN 1.0 ELSE 0 END)
            / COUNT(*) * 100,
            1
        )                                     AS success_rate_pct
    FROM transactions
    GROUP BY user_id
),

-- ─────────────────────────────────────────────────────────
-- CTE 3: Event Diversity (Product Breadth)
-- Metrics: Unique feature events triggered per user
-- ─────────────────────────────────────────────────────────
event_diversity AS (
    SELECT
        user_id,
        COUNT(DISTINCT event_name)            AS unique_events_used,
        COUNT(*)                              AS total_events
    FROM events
    GROUP BY user_id
),

-- ─────────────────────────────────────────────────────────
-- CTE 4: Weighted Health Component Scoring (0–100 Scale)
-- Components: Engagement (0–40), Recency (0–30), Transaction (0–30)
-- ─────────────────────────────────────────────────────────
user_health_raw AS (
    SELECT
        u.user_id,
        u.plan,
        u.city,
        u.channel,

        -- Engagement Score (0–40 Max Points)
        ROUND(
            MIN(
                COALESCE(sm.total_sessions, 0) * 1.5,
                40.0
            ),
            1
        )                                     AS engagement_score,

        -- Recency Score (0–30 Max Points: Decay based on inactivity days)
        ROUND(
            CASE
                WHEN sm.days_since_active IS NULL THEN 0.0
                WHEN sm.days_since_active <= 7   THEN 30.0
                WHEN sm.days_since_active <= 14  THEN 22.0
                WHEN sm.days_since_active <= 30  THEN 15.0
                WHEN sm.days_since_active <= 60  THEN 8.0
                ELSE 0.0
            END,
            1
        )                                     AS recency_score,

        -- Transaction Score (0–30 Max Points)
        ROUND(
            MIN(
                COALESCE(tm.successful_trx, 0) * 3.0,
                30.0
            ),
            1
        )                                     AS transaction_score,

        -- Underlying raw metrics
        COALESCE(sm.total_sessions, 0)        AS total_sessions,
        COALESCE(sm.days_since_active, 999)   AS days_inactive,
        COALESCE(tm.successful_trx, 0)        AS successful_trx,
        COALESCE(tm.total_volume_millions, 0) AS total_volume_millions,
        COALESCE(ed.unique_events_used, 0)    AS feature_breadth

    FROM users AS u
    LEFT JOIN session_metrics     AS sm ON u.user_id = sm.user_id
    LEFT JOIN transaction_metrics AS tm ON u.user_id = tm.user_id
    LEFT JOIN event_diversity     AS ed ON u.user_id = ed.user_id
),

-- ─────────────────────────────────────────────────────────
-- CTE 5: Composite Score Consolidation & Categorical Segmentation
-- ─────────────────────────────────────────────────────────
final_health AS (
    SELECT
        *,
        ROUND(
            engagement_score + recency_score + transaction_score,
            1
        )                                     AS health_score,
        CASE
            WHEN (engagement_score + recency_score + transaction_score) >= 60.0
                THEN 'Healthy'
            WHEN (engagement_score + recency_score + transaction_score) >= 35.0
                THEN 'Moderate'
            WHEN (engagement_score + recency_score + transaction_score) >= 15.0
                THEN 'At Risk'
            ELSE 'Churned'
        END                                   AS health_segment
    FROM user_health_raw
)

-- ─────────────────────────────────────────────────────────
-- FINAL OUTPUT: User Health Table
-- ─────────────────────────────────────────────────────────
SELECT
    user_id,
    plan,
    city,
    health_score,
    health_segment,
    engagement_score,
    recency_score,
    transaction_score,
    total_sessions,
    CAST(days_inactive AS INT) AS days_inactive,
    successful_trx,
    ROUND(total_volume_millions, 2) AS total_volume_millions,
    feature_breadth
FROM final_health
ORDER BY health_score DESC;
"""

df_health = sql(query_profesional, show_rows=10)

print("\n  User Health Segment Distribution:")
health_dist = df_health["health_segment"].value_counts()
for seg, n in health_dist.items():
    pct = n / len(df_health) * 100
    bar = "█" * int(pct / 4)
    print(f"  {seg:<12}: {n:4d} ({pct:5.1f}%) {bar}")
````

````python
# ============================================================
# CELL 13 — Behavioral Analytics Executive Summary Report
# Objective: Consolidate core engagement, financial performance, user tier mix,
# and strategic product recommendations into a final executive summary.
# ============================================================

# Aggregate summary metrics
q_summary = """
WITH
monthly AS (
    SELECT
        STRFTIME('%Y-%m', transaction_date)          AS month,
        SUM(CASE WHEN status = 'success'
                 THEN amount_millions ELSE 0 END)   AS rev
    FROM transactions
    GROUP BY month
),
prev_curr AS (
    SELECT
        month,
        rev,
        LAG(rev, 1) OVER (ORDER BY month ASC)        AS prev_rev
    FROM monthly
),
session_stats AS (
    SELECT
        ROUND(AVG(duration_seconds) / 60.0, 1)       AS avg_min,
        COUNT(DISTINCT user_id)                      AS active_users
    FROM sessions
)
SELECT
    (SELECT MAX(rev) FROM monthly)                    AS peak_revenue,
    (SELECT rev FROM prev_curr
     WHERE month = (SELECT MAX(month) FROM monthly))  AS last_rev,
    (SELECT prev_rev FROM prev_curr
     WHERE month = (SELECT MAX(month) FROM monthly))  AS prev_rev,
    ss.avg_min,
    ss.active_users,
    (SELECT COUNT(*) FROM users WHERE plan = 'premium') AS premium_users,
    (SELECT COUNT(*) FROM users)                       AS total_users
FROM session_stats AS ss;
"""

summary = sql(q_summary, show_rows=0, verbose=False)
s = summary.iloc[0]

mom_rev_growth = (
    (s["last_rev"] - s["prev_rev"])
    / s["prev_rev"] * 100.0
    if s["prev_rev"] else 0.0
)

print("=" * 65)
print("     BEHAVIORAL ANALYTICS REPORT — GajiKu Q1 2024")
print("=" * 65)
print(f"""
  ┌─────────────────────────────────────────────────────────┐
  │              EXECUTIVE SUMMARY & INSIGHTS               │
  ├─────────────────────────────────────────────────────────┤
  │                                                         │
  │  USER ENGAGEMENT                                        │
  │  Active Users    : {int(s['active_users']):>6,}                          │
  │  Avg Session     : {s['avg_min']:>5.1f} mins                     │
  │                                                         │
  │  FINANCIAL PERFORMANCE                                  │
  │  Peak Revenue    : Rp {s['peak_revenue']:>6.1f}M                   │
  │  MoM Growth      : {mom_rev_growth:>+5.1f}%                          │
  │                                                         │
  │  USER BASE MIX                                          │
  │  Total Users     : {int(s['total_users']):>6,}                          │
  │  Premium Users   : {int(s['premium_users']):>6,}                          │
  │  Premium Rate    : {s['premium_users']/s['total_users']*100:>5.1f}%                          │
  │                                                         │
  │  KEY STRATEGIC FINDINGS                                 │
  │  1. Primary conversion drop-off occurs at Step 2       │
  │     → Action: Resolve friction during amount selection │
  │                                                         │
  │  2. Premium users demonstrate significantly higher D30  │
  │     retention rate.                                     │
  │     → Action: Deploy upgrade prompts at Day-7 to boost LTV│
  │                                                         │
  │  3. Referral channels yield the highest RFM loyalty    │
  │     → Action: Scale referral incentives & budget       │
  │                                                         │
  │  4. 'help_open' represents the primary dead-end event  │
  │     → Action: Optimize contextual help UI at friction points│
  │                                                         │
  └─────────────────────────────────────────────────────────┘
""")
````
